# Análise dos resultados — Interferência Co-canal em Redes Wi-Fi

Este notebook reproduz a análise apresentada na **Seção 4 (Resultados e Discussão)** do artigo, a partir dos dados brutos gerados pelas 240 execuções de simulação em `scripts/dados.csv` (6 distâncias × 2 cenários de canal × 10 repetições × 2 fluxos por execução).

Ele gera:
- **Tabela 2**: comparação geral entre o Cenário A (canais ortogonais) e o Cenário B (mesmo canal).
- **Tabela 3**: métricas médias por distância, com desvio padrão.
- **Figuras 1–3**: throughput, atraso e taxa de perda de pacotes em função da distância entre AP1 e AP2, salvas em `graficos/`.

## 0. Setup e carregamento dos dados

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ajuste este caminho caso o notebook seja executado de outro diretório
CSV_PATH = Path("dados.csv")
OUTDIR = Path.cwd() / "graficos"
OUTDIR.mkdir(exist_ok=True)

df = pd.read_csv(CSV_PATH)
print(f"{len(df)} linhas carregadas ({df['Distancia'].nunique()} distâncias, "
      f"{df['Run'].nunique()} runs, {df['FlowId'].nunique()} fluxos por execução)")
df.head()
```

## 4.1. Comparação geral entre os cenários

Médias agregando todas as distâncias testadas (n = 120 fluxos por cenário: 6 distâncias × 10 runs × 2 fluxos).

```python
geral = df.groupby("MesmoCanal").agg(
    Throughput_medio_Mbps=("ThroughputMbps", "mean"),
    Atraso_medio_ms=("AtrasoMedioMs", "mean"),
    Taxa_perda_pct=("TaxaPerda", "mean"),
    n=("ThroughputMbps", "count"),
).rename(index={0: "Cenário A (ortogonal)", 1: "Cenário B (mesmo canal)"})

geral
```

```python
thr_a, thr_b = geral.loc["Cenário A (ortogonal)", "Throughput_medio_Mbps"], geral.loc["Cenário B (mesmo canal)", "Throughput_medio_Mbps"]
perda_a, perda_b = geral.loc["Cenário A (ortogonal)", "Taxa_perda_pct"], geral.loc["Cenário B (mesmo canal)", "Taxa_perda_pct"]
atraso_a, atraso_b = geral.loc["Cenário A (ortogonal)", "Atraso_medio_ms"], geral.loc["Cenário B (mesmo canal)", "Atraso_medio_ms"]

print(f"Redução de throughput:        {100*(thr_a-thr_b)/thr_a:.1f}%")
print(f"Aumento de perda de pacotes:  {perda_b - perda_a:.2f} p.p. (de {perda_a:.2f}% para {perda_b:.2f}%)")
print(f"Aumento de atraso:            {atraso_b/atraso_a:.0f}x (de {atraso_a:.4f}ms para {atraso_b:.2f}ms)")
```

## 4.2. Efeito da distância entre os Access Points

Médias e desvio padrão por distância e cenário (n = 20 fluxos por ponto: 10 runs × 2 fluxos).

```python
por_distancia = df.groupby(["Distancia", "MesmoCanal"]).agg(
    Throughput_mean=("ThroughputMbps", "mean"), Throughput_std=("ThroughputMbps", "std"),
    Atraso_mean=("AtrasoMedioMs", "mean"), Atraso_std=("AtrasoMedioMs", "std"),
    Perda_mean=("TaxaPerda", "mean"), Perda_std=("TaxaPerda", "std"),
).reset_index()

por_distancia
```

```python
# tabela pivotada de redução percentual de throughput por distância (Cenário A vs B)
piv = df.pivot_table(index="Distancia", columns="MesmoCanal", values="ThroughputMbps", aggfunc="mean")
piv.columns = ["Cenário A (ortogonal)", "Cenário B (mesmo canal)"]
piv["Reducao_pct"] = 100 * (piv["Cenário A (ortogonal)"] - piv["Cenário B (mesmo canal)"]) / piv["Cenário A (ortogonal)"]
piv.round(2)
```

## Figuras 1–3: métricas em função da distância

Cada figura compara os dois cenários (linha verde = canais ortogonais / controle, linha vermelha = mesmo canal / interferência), com barras de desvio padrão, e é salva em `graficos/`.

```python
dists = sorted(df["Distancia"].unique())
g0 = por_distancia[por_distancia.MesmoCanal == 0].set_index("Distancia").reindex(dists)
g1 = por_distancia[por_distancia.MesmoCanal == 1].set_index("Distancia").reindex(dists)

plt.rcParams.update({"font.size": 12, "figure.dpi": 150})


def plot_metric(mean0, std0, mean1, std1, ylabel, title, fname):
    fig, ax = plt.subplots(figsize=(7, 4.5))
    ax.errorbar(dists, mean0, yerr=std0, marker="o", capsize=4,
                 label="Canais ortogonais (sem interferência)", color="#2E7D32", linewidth=2)
    ax.errorbar(dists, mean1, yerr=std1, marker="s", capsize=4,
                 label="Mesmo canal (interferência co-canal)", color="#C62828", linewidth=2)
    ax.set_xlabel("Distância entre AP1 e AP2 (m)")
    ax.set_ylabel(ylabel)
    ax.set_title(title)
    ax.set_xticks(dists)
    ax.legend()
    ax.grid(True, linestyle="--", alpha=0.4)
    fig.tight_layout()
    fig.savefig(OUTDIR / fname, bbox_inches="tight")
    plt.show()
```

```python
# Figura 1 — Throughput
plot_metric(g0.Throughput_mean, g0.Throughput_std, g1.Throughput_mean, g1.Throughput_std,
            "Throughput médio (Mbps)", "Throughput em função da distância entre APs",
            "throughput_vs_distancia.png")
```

```python
# Figura 2 — Atraso médio
plot_metric(g0.Atraso_mean, g0.Atraso_std, g1.Atraso_mean, g1.Atraso_std,
            "Atraso médio (ms)", "Atraso médio em função da distância entre APs",
            "atraso_vs_distancia.png")
```

```python
# Figura 3 — Taxa de perda de pacotes
plot_metric(g0.Perda_mean, g0.Perda_std, g1.Perda_mean, g1.Perda_std,
            "Taxa de perda de pacotes (%)", "Taxa de perda de pacotes em função da distância entre APs",
            "perda_vs_distancia.png")
```

## 4.3. Observação — efeito de platô a partir de 5 m

As três métricas do Cenário B (mesmo canal) melhoram entre 1 m e 5 m, mas **estabilizam** a partir de 5 m — os valores em 5 m, 7 m e 10 m são praticamente idênticos. Isso indica que, com os parâmetros padrão do ns-3 usados na simulação (potência de transmissão e modelo de perda de propagação não configurados explicitamente), a interferência co-canal **não desaparece** dentro do intervalo de distâncias testado. Ver a discussão completa na Seção 4.3 do artigo (`secao-4-resultados-e-discussao.md`).
